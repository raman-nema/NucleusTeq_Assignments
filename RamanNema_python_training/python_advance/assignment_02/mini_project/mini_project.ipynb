```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Creating student dataset
student_data = {
    "Name": ["Rahul", "Priya", "Siri", "Anuj"],
    "Marks": [70, 80, 90, 60],
    "Hours_Studied": [2, 3, 5, 1]
}

# Creating DataFrame
students_df = pd.DataFrame(student_data)

# Displaying DataFrame
print(students_df)
```

```python
# Adding Performance column
# .apply() executes a function on every value in the Marks column.
students_df["Performance"] = students_df["Marks"].apply(   
    lambda marks: "Pass" if marks > 65 else "Fail"
)

print(students_df)
```

```python
# Creating line chart
plt.figure(figsize=(6, 4))

plt.plot(
    students_df["Hours_Studied"],
    students_df["Marks"],
    marker="o"
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python

# Creating scatter plot
plt.figure(figsize=(6, 4))

plt.scatter(
    students_df["Hours_Studied"],
    students_df["Marks"]
)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
# Creating barplot
plt.figure(figsize=(6, 4))

sns.barplot(
    data=students_df,
    x="Performance",
    y="Marks"
)

plt.title("Performance vs Marks")

plt.show()
```
