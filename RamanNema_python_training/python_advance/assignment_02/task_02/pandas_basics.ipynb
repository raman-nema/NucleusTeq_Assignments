```python
# Importing pandas library 
import pandas as pd
```

```python
# Creating employee data
employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj"],
    "Age": [25, 30, 28, 35],
    "Department": ["HR", "IT", "Finance", "IT"],
    "Salary": [30000, 50000, 45000, 60000]
}

# Creating the DataFrame
employees_df = pd.DataFrame(employee_data)

# Displaying DataFrame
print(employees_df)
```

```python
# Display first 2 rows of the DataFrame using head(2)
first_two_rows = employees_df.head(2)

print(first_two_rows)
```

```python
# Display summary statistics for numerical columns using describe()
summary_statistics = employees_df.describe()

print(summary_statistics)
```

```python
# Filter employees belonging to IT department
it_employees = employees_df[employees_df["Department"] == "IT"]

print(it_employees)
```

```python
# Adding Bonus column as 10% of Salary
employees_df["Bonus"] = employees_df["Salary"] * 0.10

print(employees_df)
```
