```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Creating employee dataset with missing values
employee_data = {
    "Name": ["Rahul", "Priya", "Anuj"],
    "Age": [25, np.nan, 29],
    "Salary": [30000, 40000, np.nan]
}

# Creating DataFrame
employees_df = pd.DataFrame(employee_data)

# Displaying DataFrame
print(employees_df)
```

```python
# Check for missing values in each column using isnull()
# isnull() -> true for null and false for non null values.
missing_values = employees_df.isnull()

print(missing_values)
```

```python
# Counting missing values column-wise
print(employees_df.isnull().sum())
```

```python
# Calculating mean age
mean_age = employees_df["Age"].mean()

# Replacing missing Age values with mean using fillna()
employees_df["Age"] = employees_df["Age"].fillna(mean_age)

print(employees_df)
```

```python
# Replacing missing Salary values with 0 using same fillna()
employees_df["Salary"] = employees_df["Salary"].fillna(0)

print(employees_df)
```
