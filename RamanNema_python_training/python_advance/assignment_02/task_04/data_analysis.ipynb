```python
# Import pandas library
import pandas as pd
```

```python
# Creating employee dataset
employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj", "Raman", "Naman"],
    "Age": [25, 30, 28, 35, 22,26],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance"],
    "Salary": [30000, 50000, 45000, 60000, 50000, 54000]
}

# Creating DataFrame
employees_df = pd.DataFrame(employee_data)

# Displaying DataFrame
print(employees_df)
```

```python
# Grouping data by department and calculating average salary
average_salary_by_department = employees_df.groupby("Department")["Salary"].mean()

print("Average Salary by Department:")
print(average_salary_by_department)
```

```python
# Group data by department and finding maximum salary
maximum_salary_by_department = employees_df.groupby("Department")["Salary"].max()

print("Maximum Salary by Department:")
print(maximum_salary_by_department)
```

```python
# Group data by department and count employees
employee_count_by_department = employees_df.groupby("Department")["Name"].count()

print("Employee Count by Department:")
print(employee_count_by_department)
```
