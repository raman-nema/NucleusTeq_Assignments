```python
# Import matplotlib library
import matplotlib.pyplot as plt
```

```python
# Department-wise employee count
departments = ["HR", "IT", "Finance"]
employees = [5, 12, 7]

# Salary data for histogram
salaries = [30000, 40000, 50000, 60000, 45000]

# Age and salary data for scatter plot
ages = [25, 30, 28, 35]
employee_salaries = [30000, 50000, 45000, 60000]
```

```python
# Creating a bar chart
plt.figure(figsize=(6, 4))
plt.bar(departments, employees)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Creating a line chart
plt.figure(figsize=(6, 4))
plt.plot(departments, employees, marker="o")

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

```python
# Creating a histogram for salaries
plt.figure(figsize=(6, 4))
plt.hist(salaries)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
# Creating a  scatter plot for Age vs Salary
plt.figure(figsize=(6, 4))
plt.scatter(ages, employee_salaries)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
