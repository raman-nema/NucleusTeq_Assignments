```python
# Importing required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Creating employee dataset
employee_data = {
    "Name": ["Rahul", "Priya", "Amit", "Anuj", "Raman", "Naman"],
    "Age": [25, 30, 28, 35, 22, 26],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance"],
    "Salary": [30000, 50000, 45000, 60000, 50000, 54000]
}

# Create DataFrame
employees_df = pd.DataFrame(employee_data)

# Display DataFrame
print(employees_df)
```

```python
# Creating a  barplot for Department vs Salary
plt.figure(figsize=(6, 4))

sns.barplot(
    data=employees_df,
    x="Department",
    y="Salary"
)

plt.title("Department vs Salary")
plt.show()
```

```python
# Creating boxplot for salary distribution
plt.figure(figsize=(6, 4))

sns.boxplot(
    y=employees_df["Salary"]
)

plt.title("Salary Distribution")
plt.show()
```

```python
# Calculating correlation matrix
correlation_matrix = employees_df[["Age", "Salary"]].corr()

# Creating heatmap
plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation_matrix,
    annot=True
)

plt.title("Correlation Between Age and Salary")
plt.show()
```
